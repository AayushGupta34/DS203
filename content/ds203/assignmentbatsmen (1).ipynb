```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact
```

```python
bat1= np.loadtxt('../slides/batsmen/dhoni.txt')
bat2=np.loadtxt('../slides/batsmen/kohli.txt')
bat3=np.loadtxt('../slides/batsmen/sachin.txt')
```

```python
def plotscores(typeofscore):
    if typeofscore=="boxplot":
        plt.boxplot((bat1,bat2,bat3));
    else:
        plt.violinplot((bat1,bat2,bat3));
    
```

```python
interact(plotscores, typeofscore=["boxplot", "violin"])
```

```python
batsmen = [("Dhoni", bat1), ("Kohli", bat2), ("Sachin", bat3)]
for name,scores in batsmen:
    totalgames=len(scores)
    averagevalue=np.mean(scores)
    pzero= np.sum(scores==0)/totalgames
    p50plus=np.sum(scores>=50)/totalgames
    stdev = np.std(scores)
    print(f"Name of player: {name}")
    print(f"Expected Value E[X]: {averagevalue:.2f} runs")
    print(f"P(X = 0): {pzero*100:.2f}%")
    print(f"P(X >= 50) [50+ Rate]: {p50plus*100:.2f}%")
    print(f"Standard Deviation σ : {stdev:.2f}\n")
```

```python
#Based on the mean score and the probability of the player making more than 50 runs, Kohli is the best player out of the three chosen.
#Dhoni is more consistent with his scores (least standard deviation)
```
